## Introducing the Titanic dataset

- 891 passenger records in the training set
- Features include: Passenger class (Pclass), Sex, Age, Siblings/Spouses aboard (SibSp), Parents/Children aboard (Parch), Fare, and Port of Embarkation (Embarked)

## Machine learning on the Titanic dataset

- Framed as a **supervised binary classification** problem: Predict whether a passenger survived (1) or did not survive (0)
- Famous baseline dataset for binary classification and handling real-world tabular data
- Learn more about the dataset: [Kaggle Titanic Competition](https://www.kaggle.com/c/titanic)

## Loading the Titanic dataset

```python
# import pandas and seaborn to load and inspect the dataset
import pandas as pd
import seaborn as sns
```

```python
# load the titanic dataset into a pandas DataFrame
titanic = sns.load_dataset('titanic')
type(titanic)
```

```python
# inspect the first few rows of the raw dataset
titanic.head()
```

## Machine learning terminology

- Each row is an **observation** (also known as: sample, example, instance, record)
- Each column is a **feature** (also known as: predictor, attribute, independent variable, input, regressor, covariate)

```python
# print all column names in the dataset
print(list(titanic.columns))
```

```python
# print the target values (0 = did not survive, 1 = survived)
print(titanic['survived'].values)
```

```python
# print the target class labels
print(['Did not survive (0)', 'Survived (1)'])
```

- Each value we are predicting is the **response** (also known as: target, outcome, label, dependent variable)
- **Classification** is supervised learning in which the response is categorical (binary: survived vs. died)
- **Regression** is supervised learning in which the response is ordered and continuous

## Requirements for working with data in scikit-learn

1. Features and response are **separate objects**
2. Features and response should be **numeric** (impute missing values & encode categorical variables)
3. Features and response should be **NumPy arrays** (or clean DataFrames converted to arrays)
4. Features and response should have **specific shapes**

```python
# prepare a subset of clean numeric features for modeling
features = ['pclass', 'age', 'sibsp', 'parch', 'fare']
df_subset = titanic[features + ['survived']].dropna()

# separate features (X) and response (y) into NumPy arrays
X = df_subset[features].values
y = df_subset['survived'].values

# check the types of the features and response
print(type(X))
print(type(y))
```
